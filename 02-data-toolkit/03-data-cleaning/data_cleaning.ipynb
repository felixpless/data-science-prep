```python
import pandas as pd
import numpy as np

data = {
    "Shipment_ID": [1001, 1002, 1003, 1004, 1005, 1005, 1006, 1007],
    "Customer": [
        "Amazon",
        "Zalando",
        "amazon",
        "Otto",
        "Amazon",
        "Amazon",
        "Zalando",
        "OTTO"
    ],
    "Revenue": [1200, 850, 1500, np.nan, 2100, 2100, 950, 1100],
    "Delivery_Days": [2, 4, 3, 1, np.nan, np.nan, 2, 3],
    "Status": [
        "Delivered",
        "Delivered",
        "Delayed",
        "Delivered",
        "delayed",
        "delayed",
        "Delivered",
        "Delivered"
    ]
}

shipments = pd.DataFrame(data)

shipments
```

```python
shipments["Revenue"].isna()
```

### Exercise 02.03.01 — Detect Missing Values

```python
shipments.isna().sum()
```

### Exercise 02.03.02 — Handling Missing Values

```python
median_delivery = shipments["Delivery_Days"].median()
```

```python
shipments["Delivery_Days"] = shipments["Delivery_Days"].fillna(median_delivery)
shipments
```

```python
shipments.isna().sum()
```

```python
shipments[shipments["Delivery_Days"] == 2.5]
```

### Exercise 02.03.04 — Handling Missing Revenue

```python
missing_revenue = shipments[shipments["Revenue"].isna()]
missing_revenue
```

### Exercise 02.03.05 — Detecting Duplicate Rec

```python
shipments.duplicated()
```

```python
shipments.duplicated(subset=["Shipment_ID"])
```

```python
duplicated_id = shipments[shipments["Shipment_ID"].duplicated()]
duplicated_id
```

### Exercise 02.03.06 — Removing Duplicates

```python
shipments = shipments.drop_duplicates(subset=["Shipment_ID"])

print(shipments)
print(shipments.shape[0])
```

### Exercise 02.03.07 — Standardizing Customer Names

```python
shipments
```

```python
shipments["Customer"].value_counts()
```

```python
shipments["Customer"] = shipments["Customer"].str.title()

shipments["Customer"].value_counts()
```

### Exercise 02.03.08 — Cleaning Whitespace and Text Formatting

```python
shipments.loc[0, "Customer"] = " Amazon "
shipments.loc[1, "Customer"] = " zalando"
shipments.loc[3, "Customer"] = "OTTO "
```

```python
shipments["Customer"].tolist()
```

```python
shipments["Customer"] = shipments["Customer"].str.strip().str.title()
```

```python
shipments["Customer"].value_counts()
```

### Exercise 02.03.09 — Correcting Data Types

```python
shipments["Revenue"] = shipments["Revenue"].astype("string")

shipments.dtypes
```

```python
shipments
```

```python
shipments["Revenue"] = shipments["Revenue"].astype("float")
shipments["Revenue"].dtype
```

```python
shipments["Revenue"] = pd.to_numeric(
    shipments["Revenue"],
    errors="coerce"
)

shipments["Revenue"]
```

## Exercise 02.03.10 — Final Data Cleaning Challenge
Today, you’ll apply everything you’ve learned to a realistic logistics dataset.

Your role: Data Analyst at a logistics company.

Business scenario: Your manager has received a shipment report from several operational systems. Before the data can be used for financial reporting and operational KPIs, you need to ensure its quality.

Unlike our previous exercises, I won’t provide the complete solution upfront. You’ll develop the cleaning process yourself, one task at a time.

```python
import pandas as pd
import numpy as np

raw_data = {
    "Shipment_ID": [
        2001, 2002, 2003, 2004, 2005,
        2005, 2006, 2007, 2008, 2009
    ],
    "Customer": [
        "Amazon",
        " zalando ",
        "AMAZON",
        "Otto",
        "Zalando",
        "Zalando",
        " otto ",
        "Amazon ",
        "OTTO",
        "zalando"
    ],
    "Revenue": [
        "1200", "850", "unknown", "650", "2100",
        "2100", "950", "1750", None, "1100"
    ],
    "Delivery_Days": [
        2, 4, 3, np.nan, 5,
        5, 2, np.nan, 3, 4
    ],
    "Status": [
        "Delivered",
        "delivered",
        "Delayed",
        "Delivered",
        "Delayed",
        "Delayed",
        " delivered ",
        "DELAYED",
        "Delivered",
        "delivered"
    ]
}

raw_shipments = pd.DataFrame(raw_data)

raw_shipments
```

## Data Cleaning Challenge: Shipments

Your objective is to produce a cleaned DataFrame called `cleaned_shipments`.

The final dataset should satisfy the following requirements:

| Requirement | Expected Result |
|---|---|
| Preserve the original dataset | `raw_shipments` remains unchanged |
| Remove duplicate shipment IDs | 9 unique shipments |
| Standardize customer names | `Amazon`, `Zalando`, `Otto` |
| Standardize shipment statuses | `Delivered`, `Delayed` |
| Convert revenue to numerical values | Invalid values become `NaN` |
| Handle missing delivery times | Replace with median |
| Preserve unknown revenue | Do not invent financial values |
| Verify data quality | Check missing values, duplicates, and data types |

### Important

**Do not overwrite `raw_shipments`.**

We want to preserve the original dataset so we can compare it with the cleaned version.

```python
cleaned_shipments = raw_shipments.copy()
```

```python
cleaned_shipments.isna().sum()
```

```python
cleaned_shipments
```

```python
cleaned_shipments.duplicated
cleaned_shipments = cleaned_shipments.drop_duplicates(subset=["Shipment_ID"])
```

```python
median_delivery = cleaned_shipments["Delivery_Days"].median()
cleaned_shipments["Delivery_Days"] = cleaned_shipments["Delivery_Days"].fillna(median_delivery)
```

```python
cleaned_shipments["Customer"]
```

```python
cleaned_shipments["Customer"] = cleaned_shipments["Customer"].str.title().str.strip()
```

```python
cleaned_shipments
```

```python
cleaned_shipments["Status"] = cleaned_shipments["Status"].str.title().str.strip()
```

```python
cleaned_shipments["Revenue"] = pd.to_numeric(cleaned_shipments["Revenue"],errors="coerce")
```
