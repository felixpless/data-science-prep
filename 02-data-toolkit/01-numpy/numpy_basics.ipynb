Broadcasting

```python
revenues = np.array([1000, 2000, 3000])

cost_rates = np.array([0.20, 0.30, 0.25])
```

```python
operating_costs = revenues * cost_rates

profits = revenues - operating_costs

profit_margin = profits / revenues * 100

print(f"Operating Costs are {operating_costs}")
print(f"Profits are {profits}")
print(f"Profit Margin is {profit_margin}")
```

```python
import numpy as np

# Rows: Berlin, Hamburg, Munich, Cologne
# Columns: January, February, March

revenues = np.array([
    [12000, 13500, 15000],
    [18000, 19500, 21000],
    [14000, 15500, 17000],
    [10000, 11500, 13000]
])

cost_rates = np.array([0.25, 0.30, 0.20, 0.35])
```

Adjusting Arrays with Numpy

```python
print(revenues.shape)
print(cost_rates.shape)
adjusted_cost_rates = cost_rates.reshape(4,1)
print(adjusted_cost_rates.shape)
```

Operating costs for each city and month

```python
ops_cost = revenues * adjusted_cost_rates
print(ops_cost)
```

Total quarterly profit per city

```python
profit_quarterly = revenues - ops_cost

quarterly_profits = profit_quarterly.sum(axis=1)

print(quarterly_profits)
```

Average monthly revenue per city

```python
avg_revenue_city = revenues.mean(axis=1)
print(avg_revenue_city)
```

Total company profit across all cities and months

```python
total_profit = profit_quarterly.sum()
print(total_profit)
```

Number of cities with quarterly profit above €30,000

```python

high_profit_cities = (quarterly_profits > 30000).sum()

print(high_profit_cities)
```
