### Identify High-Value Delayed Shipments

```python
delayed_high_shipments = shipments[
    (shipments["Revenue"] > 1000) &
    (shipments["Status"] == "Delayed")
]
delayed_high_shipments
```

```python
delayed_high_shipments.shape
```

```python
delayed_high_shipments.shape[0]
```

### AND vs. OR

```python
delayed_more_than_4 = shipments[
    (shipments["Status"] == "Delayed") |
    (shipments["Delivery_Days"] > 4)
]
```

```python
delayed_more_than_4.shape[0]
```

### Sorting & Ranking

```python
shipments.sort_values(by="Revenue", ascending=False).head()
```

```python
shipments[["Shipment_ID","Customer","Revenue"]].sort_values(by="Revenue", ascending=False).head()
```

```python
shipments.sort_values(by="Revenue", ascending=False).head()[["Shipment_ID","Customer","Revenue"]]
```

### Creating Calculated Columns

```python
shipments["Operating_Costs"] = shipments["Revenue"] * 0.30
```

```python
shipments.sort_values(by = "Operating_Costs", ascending = False)
```

```python
shipments["Profit"] = shipments["Revenue"] - shipments["Operating_Costs"]
shipments["Profit_Margin"] = shipments["Profit"] / shipments["Revenue"] *100
```

```python
shipments[["Profit","Profit_Margin"]].sort_values(by = "Profit", ascending = False).head()
```

### Data Types & Dates
