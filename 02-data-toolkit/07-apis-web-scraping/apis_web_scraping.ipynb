## Exercise 02.07.01A — Your First API Call

```python
import requests
import pandas as pd

url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url, timeout=10)
data_json = response.json()
print(data_json)
print(response.status_code)
print(type(data_json))
print(len(data_json))
print(data_json[0]["name"])
print(data_json[0]["email"])
```

## 02.07.01B — From API Response to Pandas DataFrame

```python
customer_df = pd.json_normalize(data_json)
print(customer_df.head())
print(customer_df.shape)
print(customer_df.columns)
```

```python
customer_contacts = customer_df.copy()
customer_contacts = customer_contacts.drop(
    columns=[
        "username","phone","website","address.street","address.suite","address.zipcode","address.geo.lat","address.geo.lng","company.catchPhrase","company.bs"
    ])
customer_contacts = customer_contacts.rename(columns={
    "address.city":"City",
    "company.name":"Compnay"
    })
print(f"Number of unique Citys: {customer_contacts["City"].nunique()}")
print(f"Customers living in the same city: {customer_contacts["City"].value_counts()}")
```

## 02.07.02 — API Query Parameters & Filtering

## Exercise 02.07.02A — Filtering API Responses

```python
import pandas
import requests 

url = "https://jsonplaceholder.typicode.com/users"

params = {"id": 3}

response = requests.get(
    url,
    params = params,
    timeout = 10
)

data = response.json()

print(response.raise_for_status())
print(response.url)
print(len(data))
print(data[0]["name"])
print(data[0]["company"]["name"])
```

## 02.07.02B — Working with Multiple API Parameters

## Exercise 02.07.02B — Multiple Query Parameters

```python
params = {
    "userId": 2,
    "id":12
}

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(
    url,
    params = params,
    timeout = 10
)

data = response.json()
response.raise_for_status()
print(response.url)
print(f"Number of returned posts: {len(data)}")
print(f"Title of returned post: {data[0]["title"]}")
```

## Exercise 02.07.02C — Handling Empty API Responses

```python
params = {
    "userId": 2,
    "id":12
}

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(
    url,
    params = params,
    timeout = 10
)
response.raise_for_status()


data = response.json()

if data:
    print(f"Title of returned post: {data[0]["title"]}")
else:
    print("No Records Found")

print(response.url)
```

## 02.07.03 — API Error Handling & Reliability

### Exercise 02.07.03A — Handling HTTP Errors

```python
url = "https://jsonplaceholder.typicode.com/invalid-endpoint"


try:
    response = requests.get(
        url,
        timeout=10
    )
    response.raise_for_status()
    data = response.json()
    print(f"Succesfully retrieved {len(data)} posts")
except requests.exceptions.HTTPError as error:
    print(f"API request failed: {error}")
    print(f"Response Status Code: {response.status_code}")
```

## 02.07.03B — Handling Timeouts & Connection Errors

```python
url = "https://jsonplaceholder.typicode.com/posts"


try:
    response = requests.get(
        url,
        timeout=5
    )
    response.raise_for_status()
    data = response.json()
    print(f"Succesfully retrieved {len(data)} posts")

except requests.exceptions.Timeout:

    print("The API request timed out.")

except requests.exceptions.ConnectionError:

    print("Could not connect to the API.")

except requests.exceptions.HTTPError:

    print("The API returned an HTTP error.")
```

## 02.07.03C — Implementing Retry Logic

## Exercise 02.07.03C — Build Your First Retry Mechanism

```python
import requests
import time


url = "https://jsonplaceholder.typicode.com/posts"

for attempt in range(3):
    print(f"Attempt {attempt +1}")


    try:
        response = requests.get(
            url,
            timeout=5
        )
        response.raise_for_status()
        data = response.json()
        print(f"Succesfully retrieved {len(data)} posts")
        break

    except requests.exceptions.Timeout:

        print("The API request timed out.")
        if attempt < 2:
            delay = 2 ** attempt
            time.sleep(delay)

    except requests.exceptions.ConnectionError:

        print("Could not connect to the API.")

    except requests.exceptions.ConnectionError:
        print("Could not connect to the API.")

    except requests.exceptions.HTTPError:

        print("The API returned an HTTP error.")
```
